```python
from ultralytics import YOLO, YOLOWorld
```

```python
# Load an Open Images Dataset V7 pretrained YOLOv8n model
model = YOLO("yolov8n-oiv7.pt")
model.predict(
    source=source,
    save=True,
    project=project,
    name="OpenImagesV7_YOLOv8n"
)

model = YOLO("yolo11n.pt")  
model.predict(
    source=source,
    save=True,
    project=project,
    name="yolov11"
)

model = YOLO("kaggle-best.pt")  
model.predict(
    source=source,
    save=True,
    project=project,
    name="HomeOjects3k"
)

model = YOLOWorld("yolov8s-worldv2.pt")
model.predict(
    source=source,
    save=True,
    project=project,
    name="yolov8s-worldv2"
)

model = YOLOWorld("yolov8s-worldv2-lvis.pt")
model.predict(
    source=source,
    save=True,
    project=project,
    name="yolov8s-worldv2-lvis"
)

model = YOLO('yolo11n_object365.pt')
model.predict(
    source=source,
    save=True,
    project=project,
    name="yolo11n_object365"
)
```
