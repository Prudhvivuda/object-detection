```python
import os
from ultralytics import YOLO
from PIL import Image
import matplotlib.pyplot as plt

# Define your test images and models
test_image_dir = "test_images"
test_images = [os.path.join(test_image_dir, f) for f in os.listdir(test_image_dir) if f.endswith((".jpg", ".png"))]

models = {
    "homeobjects3k": "yolov11n-homeobjects3k.pt",
    "lvis": "yolo8s-worldv2-lvis.pt",
    "object365": "yolov11n-object365.pt",
}

output_dir = "benchmark_outputs"
os.makedirs(output_dir, exist_ok=True)

# Loop through each model and predict on test images
for model_name, model_path in models.items():
    print(f"\n🔍 Running model: {model_name}")
    model = YOLO(model_path)
    model_output_dir = os.path.join(output_dir, model_name)
    os.makedirs(model_output_dir, exist_ok=True)

    for image_path in test_images:
        print(f"Predicting on: {os.path.basename(image_path)}")

        results = model.predict(source=image_path, save=True, project=model_output_dir, name="results", save_txt=True, imgsz=640)

        # Display one image with predictions
        pred_image_path = results[0].save_dir / results[0].path.name
        img = Image.open(pred_image_path)
        plt.imshow(img)
        plt.title(f"{model_name} - {os.path.basename(image_path)}")
        plt.axis('off')
        plt.show()
```
